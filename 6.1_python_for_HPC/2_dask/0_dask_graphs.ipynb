# Inspect a Dask task graph

**Optional deep dive, about 10 minutes.** Dask records a plan called a task graph. The plan shows which tasks are independent and which must wait for earlier tasks.

```python
from pathlib import Path
import dask.array as da

array = da.arange(15, chunks=5)
array
```

```python
print("Chunks:", array.chunks)
print("Blocks:", array.numblocks)
print("Graph keys:", list(array.dask)[:5])
```

Build an expression without calculating the answer yet. Each chunk can add one independently, then Dask adds the partial results into the final sum.

```python
result = (array + 1).sum()
try:
    result.visualize(filename="task_graph.png", optimize_graph=False, engine="graphviz")
except Exception as error:
    print("Graphviz visualization unavailable:", error)
```

```python
actual = result.compute()
expected = sum(range(1, 16))
assert actual == expected
print("Verified result:", actual)
```

Change the chunk size to 3 and 15. Before running, predict how the number of graph tasks will change. The numerical result must remain the same.

<details><summary><b>Solution & Explanation</b></summary>

Click expand below to inspect how chunk size changes block and task graph counts.
</details>

```python
for chunk_size in (3, 15):
    candidate = da.arange(15, chunks=chunk_size)
    candidate_result = (candidate + 1).sum()
    print(
        f"chunk={chunk_size}, blocks={candidate.numblocks}, "
        f"graph_tasks={len(candidate_result.dask)}, result={candidate_result.compute()}"
    )

Path("task_graph.png").unlink(missing_ok=True)
```

## Takeaway

Chunk size changes both the number of blocks and the layout of the task graph. Inspect the number of tasks before using larger data.
