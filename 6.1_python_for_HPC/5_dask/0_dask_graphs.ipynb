# Inspect a Dask task graph

**Optional deep dive, about 10 minutes.** Dask records operations in a directed acyclic graph. Dependencies determine which tasks can run at the same time.

```python
from pathlib import Path
import dask.array as da

array = da.arange(15, chunks=5)
array
```

```python
print("Chunks:", array.chunks)
print("Blocks:", array.numblocks)
print("Graph keys:", list(array.dask)[:5])
```

Build an expression without computing it. Each chunk can add one independently, then the partial sums feed a final reduction.

```python
result = (array + 1).sum()
result.visualize(filename="task_graph", format="svg", optimize_graph=False)
result
```

```python
actual = result.compute()
expected = sum(range(1, 16))
assert actual == expected
print("Verified result:", actual)
```

## Your turn

Change the chunk size to 3 and 15. Before running, predict how the number of graph tasks will change. The numerical result must remain the same.

```python
for chunk_size in (3, 15):
    candidate = da.arange(15, chunks=chunk_size)
    candidate_result = (candidate + 1).sum()
    print(
        f"chunk={chunk_size}, blocks={candidate.numblocks}, "
        f"graph_tasks={len(candidate_result.dask)}, result={candidate_result.compute()}"
    )

Path("task_graph.svg").unlink(missing_ok=True)
```
